```python
with engine.connect() as cn:
    consulta = text('''
        SELECT condition, COUNT(*) AS coches, ROUND(AVG(price)::numeric, 0) AS media_usd
        FROM anuncios
        GROUP BY condition ORDER BY coches DESC
    ''')
    print(pd.read_sql(consulta, cn))
```
